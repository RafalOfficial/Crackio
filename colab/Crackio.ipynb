# Crackio on Colab

Press the **play button** below. About 3 minutes later an **Open Crackio** button appears under it.

- **GPU:** this notebook asks for an L4 with High-RAM. Want another one? Choose **Runtime → Change runtime type** before you press play. When Colab has no L4 free it connects another GPU (often a T4): Crackio runs on that too.
- **Cost:** Crackio is free. The GPU uses your Colab compute units (Pro or Pay As You Go). Colab's free plan often stops apps like this one.
- **Your work:** everything lives on this runtime and is gone when it ends. Download what you want to keep.
- **When you're done:** **Runtime → Disconnect and delete runtime**, so it stops using units.

Colab asks once whether to run a notebook that comes from GitHub: choose **Run anyway**. The code is open: [github.com/RafalOfficial/Crackio](https://github.com/RafalOfficial/Crackio).

```python
#@title Start Crackio { display-mode: "form" }
import os, runpy, shutil, subprocess

APP = "/content/crackio"
if not os.path.exists(APP + "/colab/start.py"):
    print("Getting Crackio ...", flush=True)
    for attempt in range(3):
        shutil.rmtree(APP, ignore_errors=True)
        if subprocess.call(["git", "clone", "--quiet", "--depth", "1", "https://github.com/RafalOfficial/Crackio.git", APP]) == 0:
            break
if os.path.exists(APP + "/colab/start.py"):
    runpy.run_path(APP + "/colab/start.py", run_name="__main__")
else:
    print("Crackio couldn't be downloaded from GitHub. Press play again in a minute.")
```
